## 3. Operaciones aritméticas

Python tiene los operadores: `+`, `-`, `*`, `/` (división real), `//` (división entera), `%` (módulo), `**` (potencia).  
El módulo `math` ofrece funciones como `sqrt`, `ceil`, `floor`.

```python
# Ejercicio 3a
a, b = 23, 7
print('División:', a//b)
print('Residuo:', a%b)
print('Potencia:', f'{a**b}'[:5])
```

## 4. Strings y métodos útiles

Los strings tienen métodos como `.strip()`, `.upper()`, `.lower()`, `.replace()`, `.split()`, `.join()`.  
Los **f-strings** permiten insertar variables directamente en el texto: `f"Hola {nombre}"`.

```python
# Ejercicio 4a
cadena = "  fisica y matematica  "
paso1 = cadena.strip()
print(f"Paso 1: '{paso1}'")
paso2 = paso1.upper()
print(f"Paso 2: '{paso2}'")
paso3 = paso2.replace('Y', '&')
print(f"Paso 3: '{paso3}'")
```

## 5. Listas

Las listas son **mutables** y se crean con `[]`. Métodos principales: `.append()`, `.insert()`, `.remove()`, `.pop()`.  
Funciones útiles: `len()`, `min()`, `max()`, `sum()`, `sorted()`.

```python
# Ejercicio 5a
materias = ["Cálculo", "Física", "Programación", "Álgebra"]
materias.append("Estadística")
materias.insert(2, "Química")
materias.remove("Álgebra")
print(f"Lista final: {materias}")
print(f"Longitud: {len(materias)}")
```

```python
# Ejercicio 5b
notas = [78, 92, 65, 88, 74, 95, 61, 83]
print(f"Mínimo: {min(notas)}")
print(f"Máximo: {max(notas)}")
print(f"Suma: {sum(notas)}")
print(f"Promedio: {sum(notas)/len(notas):.2f}")
```

## 6. Diccionarios

Los diccionarios almacenan pares **clave → valor**. Se crean con `{}`.  
Para acceder de forma segura usa `.get(clave, valor_por_defecto)`.  
Para iterar sobre pares: `for clave, valor in d.items():`

```python
# Ejercicio 6a
curso_dict = {
 "nombre": "Física Atómica",
 "codigo": "101",
 "creditos": 5,
 "activo": True
}
curso_dict["estudiantes"] = 35
print(f"Créditos: {curso_dict.get('creditos')}")
print(f"Salón: {curso_dict.get('salon', 'Sin asignar')}")
```

```python
# Ejercicio 6b
for clave, valor in curso_dict.items():
    print(f"{clave:12} : {valor}")
```

## 7. Condicionales

```python
# Ejercicio 7a
def clasificar_imc(imc):
    if imc < 18.5:
        return "Bajo peso"
    elif 18.5 <= imc <= 24.9:
        return "Normal"
    elif 25 <= imc <= 29.9:
        return "Sobrepeso"
    else:
        return "Obesidad"

for v in [16.0, 22.5, 27.3, 31.8]:
    print(f"IMC {v}: {clasificar_imc(v)}")
```

```python
# Ejercicio 7b
for n in [4, 7, 0]:
    resultado = "par" if n % 2 == 0 else "impar"
    print(f"{n} es {resultado}")
```

## 9. Funciones

```python
# Ejercicio 9a
def convertir_temperatura(grados, escala="C"):
    if escala == "C":
        return grados * 9/5 + 32
    else:
        return (grados - 32) * 5/9

print(f"100C a F: {convertir_temperatura(100)}")
print(f"212F a C: {convertir_temperatura(212, 'F')}")
```

```python
# Ejercicio 9b
def resumen_lista(lista):
    return min(lista), max(lista), sum(lista)/len(lista)

mi, ma, pro = resumen_lista([12, 45, 7, 89, 34, 56, 23])
print(f"Min: {mi}, Max: {ma}, Prom: {pro:.2f}")
```

```python
# Ejercicio 9c
def fibonacci(n):
    if n <= 1: return n
    return fibonacci(n-1) + fibonacci(n-2)

print([fibonacci(i) for i in range(10)])
```

## 10. Manejo de errores

```python
# Ejercicio 10a
def convertir_entero(valor):
    try:
        return int(valor)
    except ValueError:
        print("Error de valor")
        return None
    finally:
        print(f"[intento con: {valor}]")

convertir_entero("hola")
```

```python
# Ejercicio 10b
def validar_edad(edad):
    if not isinstance(edad, int): raise TypeError("No es entero")
    if edad < 0 or edad > 150: raise ValueError("Fuera de rango")
    return True

try:
    validar_edad(200)
except Exception as e:
    print(f"Atrapado: {e}")
```

## Ejercicio integrador

```python
def analizar_curso(estudiantes):
    aprobados = 0
    reprobados = 0
    print(f"{'Nombre':20} | {'Promedio':8} | {'Estado'}")
    for est in estudiantes:
        notas = est['notas']
        promedio = sum(notas)/len(notas) if notas else None
        
        if promedio is None:
            estado = "Sin notas"
        else:
            estado = "Aprobado" if promedio >= 61 else "Reprobado"
            if estado == "Aprobado": aprobados += 1
            else: reprobados += 1
        
        p_str = f"{promedio:.2f}" if promedio is not None else "N/A"
        print(f"{est['nombre']:20} | {p_str:8} | {estado}")
    print(f"\nAprobados: {aprobados}, Reprobados: {reprobados}")

curso = [
    {"nombre": "Ana López",    "notas": [85, 90, 78, 92]},
    {"nombre": "Carlos Ruiz",  "notas": [55, 48, 60, 52]},
    {"nombre": "María Paz",    "notas": [70, 75, 68, 80]},
    {"nombre": "Luis Gómez",   "notas": [95, 98, 100, 92]},
    {"nombre": "Sara Díaz",    "notas": [40, 55, 50, 45]},
    {"nombre": "Pedro Alva",   "notas": []},
]

analizar_curso(curso)
```
