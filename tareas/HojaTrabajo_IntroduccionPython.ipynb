## 3. Operaciones aritméticas

```python
a, b = 23, 7
print('División:', a//b)
print('Residuo:', a%b)
print('Potencia:', f'{a**b}'[:5])
```

```python
import math
raiz = math.sqrt(144)
print(f'Raíz cuadrada de 144: {raiz}')
techo = math.ceil(7.3)
print(f'Techo de 7.3: {techo}')
piso = math.floor(7.9)
print(f'Piso de 7.9: {piso}')
print(f'Valor de pi: {math.pi:.4f}')
```
